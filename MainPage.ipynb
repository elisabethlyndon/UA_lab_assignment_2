## Lab 6 overview

Then next was Lab 6 for which I chose to use Bristol as a location to look into  how neighbourhoods can be classified using their demographic and socioeconomic characteristics and the potential applications of these outcomes for understanding community service requirements.

For my dataset i used Census 2021 data for 1,405 Output Areas within Bristol and then I selected 6 percentage indicators from 4 topics: age, economic activity, general health and housing tenure. After initially checking the data and then standardising the selected variables, I next applied the k-means clustering taught in lab 6 workbook and then I compared different numbers of clusters using the elbow method, silhouette scores and cluster profiles which wwre also taught in lab 6.
